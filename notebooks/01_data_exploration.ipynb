```python
import torch
import torchvision
import matplotlib.pyplot as plt

from pathlib import Path
from torchvision.datasets import GTSRB
```

```python
print("PyTorch:", torch.__version__)
print("Torchvision:", torchvision.__version__)
```

```python
PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / "data"

DATA_DIR.mkdir(exist_ok=True)

print("Dataset location:", DATA_DIR)
```

```python
DATA_DIR
```

```python
train_dataset = GTSRB(
    root=str(DATA_DIR),
    split="train",
    transform=None,
    download=True
)
```

```python
print(" train data number is "+str(len(train_dataset)))
```

```python
image,label=train_dataset[0]
print( "image type is " , type(image))
print("image size  is ",image.size)
print ("label is ", label)
```

```python
image
```

```python
plt.figure(figsize=(4, 4))

plt.imshow(image)
plt.title(f"Class label: {label}")
plt.axis("off");
```

```python
import random

plt.figure(figsize=(12, 8))

for i in range(12):
    index = random.randint(0, len(train_dataset) - 1)
    image, label = train_dataset[index]

    plt.subplot(3, 4, i + 1) ##plt.subplot() is a function in the Python Matplotlib library used to draw multiple plots in a single figure
    plt.imshow(image)
    plt.title(f"Class: {label}")
    plt.axis("off")

plt.tight_layout()
```

```python
test_dataset = GTSRB(
    root=str(DATA_DIR),
    split="test",
    download=True
)
```

```python
print(" valid data number is "+str(len(test_dataset)))
```

```python
image,label=test_dataset[0]
print( "image type is " , type(image))
print("image size  is ",image.size)
print ("label is ", label)
```

```python
import random

plt.figure(figsize=(12, 8))

for i in range(12):
    index = random.randint(0, len(test_dataset) - 1)
    image, label = test_dataset[index]

    plt.subplot(3, 4, i + 1) ##plt.subplot() is a function in the Python Matplotlib library used to draw multiple plots in a single figure
    plt.imshow(image)
    plt.title(f"Class: {label}")
    plt.axis("off")

plt.tight_layout()
```
